# MemAE on MVTec AD — experiment record

This notebook records the completed 15-category MemAE run and provides a small inspection workflow. The model implementation is in `src/memae/model.py`; the full Kaggle training run and raw artifacts are retained in Google Drive under `CS663Project/memae_runs`.

MemAE is a reconstruction-based detector: an encoder produces spatial feature vectors, a sparse memory module retrieves normality prototypes, and a decoder reconstructs the input. Pixelwise reconstruction error forms the anomaly map.


## Recorded configuration

- 256×256 RGB input; normal-only training.
- Memory size 256, hard-shrinkage threshold 0.001, temperature 0.07.
- Maximum 40 epochs, patience 8, batch size 16, learning rate 1e-3.
- Thresholds calibrated using held-out normal training images (99th image percentile; 99.5th pixel percentile).
- Seed 42.

## Interpretation and limitations

The model completed all 15 categories. Results varied by category: wood and hazelnut had strong image-level AUROC, while cable, metal nut and transistor were difficult. Training and validation reconstruction losses converged, but this did not guarantee separation of normal and defective test images. Normal reflective structures and texture could produce high reconstruction errors.

F1 depends on the normal-only percentile thresholds and should not be compared uncritically with other implementations. The maximum-pixel-score ablation was post-hoc and is not the primary reported result. Full checkpoints, maps and per-image records remain in Google Drive rather than Git.
