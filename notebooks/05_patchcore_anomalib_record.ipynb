## Interpretation and implementation notes

PatchCore builds a memory bank of normal patch embeddings from pretrained CNN features. Test patch descriptors are compared with the memory bank using nearest-neighbour distances, producing an anomaly map. Image-level scores aggregate patch-level anomaly evidence.

The recorded metrics are from Anomalib 2.7.0 and its evaluator. The local `src/patchcore/` code is an independent project implementation; its current random subset memory bank is not an exact reproduction of Anomalib PatchCore's coreset-selection pipeline. Keep that distinction in presentations and future comparisons.

The large model checkpoints, category predictions and visualisation folders are stored in Google Drive, not committed to Git. See `docs/completed_experiments_nov2026.md` for the experiment record.
